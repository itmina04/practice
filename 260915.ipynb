### 리스트에 데이터 추가
append()
insert()
extend()

```python
mynumbers.append([70, 80]) #끝에만 추가 가능, 할 때마다 늘어남
mynumbers
```

```python
mynumbers.insert(1, 15) #특정 위치(1번 자리) 지정 그 위치에 값(15)을 넣겠다
mynumbers
```

```python
mynumbers.extend([70, 80]) #리스트 없이 끝에 추가해줌
mynumbers
```

```python
#미션: 1번 인덱스 위치에 45, 55를 펼쳐서 넣게

mynumbers[1:1] = [45, 55] #시작은 포함이고, 끝은 미만이라서 빈 리스트가 나옴.
mynumbers
```

```python
print("끝말잇기 게임")
print("시작")

var = "가나다"
var_list = [var] #추가
print(var)

while True:
    answer = input("끝말잇기 단어 입력: ")
    print(answer)
    if answer == "":
        print('단어가 입력되지 않았으니 재입력 하세요')
        continue
    if answer in var_list: #추가
        print('이미 나온 단어입니다')
        break
    if var[-1] == answer[0]:
        var_list.append(answer) #추가
        var = answer #업데이트
    else:
        print('땡')
        break
    
print('게임종료')
```
