```python
def add_mul(sel, *args): #return은 여러 번 해도 되고, 중간에 넣어도 됨. 
    #단 return 이라는 키워드는 함수의 종료를 의미해서 뒷 부분은 실행하지 않고 끝남
    #print(sel, args)
    if sel == 'add':
        result = 0
        for i in args:
            result += i
    elif sel == 'mul':
        result = 1
        for i in args:
            result *= i
    else:
        print('지원하지 않는 연산자 입니다')
    return result

print(add_mul('add', 1, 2, 3, 4))
print(add_mul('mul', 1, 2, 3, 4, 5))
```

```python
def rank(score):
    avg = [sum(i)/len(i) for i in score]
    #print(avg)
    avg_sort = sorted(avg, reverse=True)
    #print(avg_sort)

    result = [avg_sort.index(i) + 1 for i in avg]
    return result

rank([[80, 70], [70, 80], [30, 50], [90, 100], [100, 90], [100, 100], [10, 30]])
```

```python
import random

def checkWin(user, com):
    if user == com:
        return "비겼습니다"

    win_cases = [('가위', '보'), ('바위', '가위'), ('보', '바위')]

    if (user, com) in win_cases:
        return "사용자님이 이겼습니다"
    else:
        return "사용자님이 졌습니다"

def continueConfirm():
    while True:
        choice = input("게임을 계속하려면 1을 눌러주세요, 그 외의 값은 게임을 종료합니다: ")
        if choice == '1':
            print('게임을 재개합니다')
            return True
        else:
            print('게임을 종료합니다')
            return False
        
sel = ['가위', '바위', '보']

while True:
    user = input('가위, 바위, 보 중 하나를 입력하세요: ')

    if user not in sel:
        print('잘못된 입력으로 게임을 종료합니다')
        break

    else:
        com = random.choice(sel)
        print(f'컴퓨터: {com}')

        result = checkWin(user, com)

        if result == '비겼습니다':
            print('비겼습니다')
        elif result == '사용자님이 이겼습니다':
            print('사용자님이 이겼습니다')
        else:
            print('사용자님이 졌습니다')

    if not continueConfirm():
        print('게임 종료')
        break
```
