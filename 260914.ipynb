# 조건문

if 조건:
    실행할 명령
elif 조건2:
    실행할 명령2
elif 조건3:
    실행할 명령3
else:
    다른 명령

조건문 밖의 명령어

```python
money = True

if money:
    print('스테이크를 먹는다')
    print('커피를 마신다')

print('조건문 밖의 명령어')
```

```python
num = 6

if num > 10:
    print('너무 커요')
elif num > 5:
    print('적당히 커요')
else:
    print('괜찮아요')
```

```python
#짝수홀수구분

num1 = 2
num2 = 3

if num1 % 2 == 0:
    print('num1은 짝수입니다.')
else:
    print('num1은 홀수입니다.')

if num2 % 2 == 0:
    print('num2은 짝수입니다.')
else:
    print('num2은 홀수입니다.')
```

```python
#성적처리프로그램
#90점 이상 -> 'A', 80점 이상 -> 'B', 70점 이상 -> 'C', 60점 이상 -> 'D', 60점 미만 -> 'F'

score = int(input("점수 입력: "))
if score >= 90:
    print('A')
elif score >= 80:
    print('B')
elif score >= 70:
    print('C')
elif score >= 60:
    print('D')
else:
    print('F')
```

# 반복문

for, while

for 변수 in 이터러블: #이터러블은 바구니 같은 역할, 횟수가 정해져 있음
    반복할 명령

바깥 명령어

```python
for i in range(5): #i는 임시변수
    print(f'Hello {i}번째 입니다')
```

```python
for i in range(1, 6, 2): #start, end, step
    print(i)
```

```python
answer = 0

for i in range(1, 11):
    answer += i

print(answer)
```

```python
dan = 2
for i in range (1, 10):
    print(f'2 * {i} = {2*i}')
```

```python
#2~9단 출력

for dan in range(2, 10):
    for i in range(1, 10):
        print(f'{dan} * {i} = {dan*i}')

```

```python
for i in 'Hello':
    print(i)
```

while 조건:
    반복할 명령

```python
for i in range(5):
    print(f'{i}번쨰 Hello')
```

```python
i = 0
while i < 5 :
    print(f'{i} 번째 Hello')
    i += 1
```

```python
input_num = 0
while input_num != 4:
    print(input_num)
    input_num = int(input('다시 입력해라: '))
```

```python
input_num = 0
while True:
    print(input_num)
    input_num = int(input('다시 입력해라: '))
    if input_num == 4:
        print('game over')
        break

print(input_num)
```

```python
#미션1.

import random

exact_num = random.randint(1, 100)

while True:
    user_num = int(input("숫자를 맞춰보세요: "))
    if user_num > exact_num:
        print(f'{user_num}은 숫자가 큽니다')
    elif user_num < exact_num:
        print(f'{user_num}은 숫자가 작습니다')
    else:
        print(f'{user_num}이 정답입니다')
        break
```
