# 클래스

```python
class Student:
    population = 0 #클래스 변수 (공유)
    
    def __init__(self, name, age): #생성자
        self.name = name #인스턴스 변수
        self.age = age
        Student.population += 1

    # 매서드
    def hello(self):
        print(f'{self.name}님 안녕하세요')
```

```python
# 상속
class NewStudent(Student):

    #새로운 매서드
    def study(self):
        print(f'{self.name}님이 공부해요')
    #매서드 오버라이딩(기존의 매서드 재정의)
    def hello(self):
        print(f'{self.name}님이 반갑게 인사해요')
```

```python
std1 = Student('최수빈', 27)
std2 = Student('최연준', 28)
std3 = NewStudent('최범규', 25)
```

```python
std3.study()
std3.hello()
```

```python
std1.hello() 
print(Student.population)
```
